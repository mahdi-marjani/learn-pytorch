# The essence of tensors

```python
points = torch.zeros(6)

print(points)

points[0] = 4.0
points[1] = 1.0
points[2] = 5.0
points[3] = 3.0
points[4] = 2.0
points[5] = 1.0

print(points)
```

```python
points = torch.tensor([4.0, 1.0, 5.0, 3.0, 2.0, 1.0])
points
```

```python
float(points[0]), float(points[1])
```

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points
```

```python
points.shape
```

```python
points = torch.zeros(3, 2)
points
```

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points
```

```python
points[0, 1]
```
